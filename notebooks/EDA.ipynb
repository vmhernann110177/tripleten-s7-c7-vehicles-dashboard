## <b><FONT SIZE=6 COLOR="#F5950F">Proyecto Sprint 7 — Análisis Exploratorio de Datos</FONT></b>
- <b><FONT SIZE=5 COLOR="#FB7BEF">Contexto del proyecto</FONT></b><br>
    - <FONT SIZE=5 COLOR="#0AACF7">Este proyecto forma parte del Sprint 7: Herramientas de desarrollo de software, del bootcamp de Ciencia de Datos de Tripleten.</FONT><br>
    - <FONT SIZE=5 COLOR="#0AACF7">El objetivo es desarrollar y desplegar una aplicación web interactiva para explorar información de anuncios de venta de vehículos, utilizando Python, pandas, Plotly y Streamlit.</FONT><br>
- <b><FONT SIZE=5 COLOR="#FB7BEF">Conjunto de datos</FONT></b><br>
    - <FONT SIZE=5 COLOR="#0AACF7">Se utilizará el archivo vehicles_us.csv, proporcionado para el proyecto, que contiene información sobre anuncios de vehículos.</FONT><br>
    - <FONT SIZE=5 COLOR="#0AACF7">Su estructura, dimensiones, tipos de datos y valores faltantes se verificarán durante el análisis exploratorio, sin asumir resultados previamente.</FONT><br>
- <b><FONT SIZE=5 COLOR="#FB7BEF">Objetivos del análisis exploratorio</FONT></b><br>
    - <FONT SIZE=5 COLOR="#0AACF7">Examinar las características generales del conjunto de datos y verificar las variables necesarias para construir visualizaciones interactivas.</FONT><br>
    - <FONT SIZE=5 COLOR="#0AACF7">Explorar la distribución del odómetro y la relación entre odómetro y precio mediante gráficos elaborados con Plotly.</FONT><br>
    - <FONT SIZE=5 COLOR="#0AACF7">Documentar los resultados observados y fundamentar la selección de visualizaciones que posteriormente se incorporarán al dashboard desarrollado con Streamlit.</FONT><br>
- <b><FONT SIZE=5 COLOR="#FB7BEF">Alcance metodológico</FONT></b><br>
    - <FONT SIZE=5 COLOR="#0AACF7">El análisis será exploratorio y estará orientado a comprender los datos y preparar las visualizaciones de la aplicación web, sin convertir esta etapa en una investigación estadística exhaustiva.</FONT><br>
    - <FONT SIZE=5 COLOR="#0AACF7">Cada bloque incluirá su explicación, código ejecutable, verificación de resultados e interpretación documentada.</FONT><br>
